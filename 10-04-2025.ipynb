```python
DF1 = pd.DataFrame({"A":[1,0,1],"B":[0,1,1]},dtype=bool)
DF2 = pd.DataFrame({"A":[1,0,0],"B":[0,1,0]},dtype=bool)
```

```python
DF1
```

```python
DF1 & DF2 #AND
```

```python
DF1 | DF2 # OR
```

```python
DF1 ^ DF2    #XOR
```

```python
-DF1      #OPPOSITE
```

```python
DF1.gt(DF2)     # > greater than
```

```python
DF1.ne(DF2)    # not equal to
```
