```python
#single inheritance
class A:
    def __init__(self):
        print("consrtuctor of A")
    def feature(self):
        print("feature of A")
    def featureone(self):
        print("featureone of A")
```

```python
class B(A):
    def __init__(self):
        super().__init__()
        print("constuctore of B")
    def featureB(self):
        print("feature  from class B")
```

```python
B1 = B()
```

```python
B1.featureB()
```

```python
B1.featureone()
```

```python
B1.feature()
```

```python
class C(B,A):
    def featureC(self):
        print("feature of C")
```
