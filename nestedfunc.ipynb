```python
import math
pi=3.14
def area(x):
    def perimeter(y):
        return 2*(x+y)*pi
    return perimeter
        
```

```python
result1 = area(4)(4)  # Calls `area` with 4, then `perimeter` with 4
print("Perimeter:", result1)  # Output: Perimeter: 50.24
```
