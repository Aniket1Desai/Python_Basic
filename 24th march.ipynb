```python
#1. Create a string  made of 1st, middle and last char/

def first_middle_last(s):
    mid = len(s) // 2             #1 question
    return s[0] + s[mid] + s[-1] if s else '' 
```

```python
print(first_middle_last("Sujal"))
```

```python
#2. create a string made of 3 middle char.

def middle_three(s):
    mid = len(s) // 2             #2 question
    return s[mid-1:mid+2] if len(s) >= 3 else s
```

```python
print(middle_three("Sujal"))
```

```python
#3. append a new string in the middle of a given string

def insert_middle(s, new):
    mid = len(s) // 2            #3 question
    return s[:mid] + new + s[mid:]
```

```python
print(insert_middle("Sujal", "Hello"))
```

```python
#4. arrange a string char in a way that lower letters should come first. 

def sort_lower_upper(s):         #4 question
    return ''.join(sorted(s, key=lambda c: (c.isupper(), c)))
```

```python
print(sort_lower_upper("SuJaL"))
```

```python
#5. count all letters, digits and special char in a given string.(ankit@1234)

def count_chars(s):             #5 question
    letters, digits, special = 0, 0, 0
    for c in s:
        if c.isalpha():
            letters += 1
        elif c.isdigit():
            digits += 1
        else:
            special += 1
    return letters, digits, special
```

```python
print(count_chars("sujal@1234"))
```

```python
#6. write a program to check if two string are balanced.   char of s1 is present in s2


def is_balanced(s1, s2):       #6 question
    return all(c in s2 for c in s1)
```

```python
print(is_balanced("abc", "cbad"))
```

```python
#7. calculate sum and avg of the digit present in a string.
    #ankit1234  1+2+3+4 aur avg(1,2,3,4)

def sum_avg_digits(s):         #7 question
    digits = [int(c) for c in s if c.isdigit()]
    return sum(digits), sum(digits) / len(digits) if digits else 0
```

```python
print(sum_avg_digits("sujal1234"))
```
